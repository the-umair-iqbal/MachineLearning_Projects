# Handwritten Digit Recognition using CNN

**Goal:** Build a simple CNN that recognizes handwritten digits **0–9**.

### Simple project flow
**Import → Load Dataset → Preprocess → Build CNN → Train → Test → Predict → Graphs**

This notebook is intentionally short so students can understand and complete the project themselves.

## 1. Import libraries

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Load the MNIST dataset
MNIST contains 28×28 grayscale images of handwritten digits from 0 to 9.

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)
```

## 3. View some images

```python
plt.figure(figsize=(8, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(y_train[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## 4. Preprocess the data

Pixel values are from **0 to 255**.  
We divide by **255** so the values become **0 to 1**.

CNN also needs a channel dimension, so we change:

**28 × 28 → 28 × 28 × 1**

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]

print(x_train.shape)
```

## 5. Build the CNN

- **Conv2D** → finds patterns
- **MaxPooling** → reduces the image size
- **Flatten** → converts features into a list
- **Dense** → makes the final decision
- **Softmax** → gives probabilities for digits 0–9

```python
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(64, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

model.summary()
```

## 6. Compile the model
**Adam** helps the model learn, **loss** measures error, and **accuracy** measures correct predictions.

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

## 7. Train the CNN
An **epoch** means one complete pass through the training data.

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)
```

## 8. Test the model
The test data was not used for training, so it checks how well the model works on unseen images.

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("Test accuracy:", test_accuracy)
```

## 9. Make predictions

```python
predictions = model.predict(x_test[:10], verbose=0)
predicted_digits = np.argmax(predictions, axis=1)

print("Predicted:", predicted_digits)
print("Actual:   ", y_test[:10])
```

## 10. Check predictions visually

```python
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i].squeeze(), cmap="gray")
    plt.title(f"Pred: {predicted_digits[i]} | Actual: {y_test[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## 11. Training vs Validation Accuracy

This graph shows whether the model is learning well.

Ideally, the **training and validation accuracy lines should stay close together**.

```python
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

## 12. Training vs Validation Loss

A lower loss generally means the model is making fewer errors.

The two lines should ideally follow a similar pattern.

```python
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(True)
plt.show()
```
